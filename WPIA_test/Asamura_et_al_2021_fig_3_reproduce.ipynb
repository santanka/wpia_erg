````python
import numpy as np
import xarray as xr
from scipy.signal import spectrogram, detrend, get_window

def spectro_to_xarray(da, fs=256.0, comp=2, nperseg=4096, noverlap=None, window='hann'):
    """
    da: DataArray(time, xyz) 例: B_256Hz
    comp: 0=Bx, 1=By, 2=Bz
    返り値: Dataset {Sxx_<comp>} with coords time(freq center), freq
    """
    if noverlap is None:
        noverlap = nperseg // 2

    x = detrend(da.isel(xyz=comp).values, type='linear')
    win = get_window(window, nperseg)

    f, t, Sxx = spectrogram(x, fs=fs, window=win, nperseg=nperseg,
                            noverlap=noverlap, detrend='linear',
                            scaling='density', mode='psd')   # Sxx: (F, T)

    # spectrogram の t は開始からの秒。絶対時刻へ変換（窓中心時刻）
    t0 = da.time.values[0].astype('datetime64[ns]')
    t_abs = t0 + (np.rint(t * 1e9).astype('int64')).astype('timedelta64[ns]')

    varname = {0:'Bx', 1:'By', 2:'Bz'}[comp]
    ds = xr.Dataset(
        data_vars={f'Sxx_{varname}': (('time', 'freq'), Sxx.T)},   # (T,F)
        coords={'time': t_abs, 'freq': f},
        attrs=dict(fs=fs, nperseg=nperseg, noverlap=noverlap, window=window,
                   input_units='nT', psd_units='nT^2/Hz',
                   method='scipy.signal.spectrogram')
    )
    return ds
````

````python
import numpy as np
from scipy.signal import butter, sosfiltfilt
import pytplot as pt
import matplotlib.pyplot as plt
import os # osモジュールもインポートしておく

# フィルタパラメータ
fs = 64.                  # サンプリング周波数 [Hz]
order = 4                 # フィルタの次数
window_sec = background_time_sec        # 背景磁場の移動平均窓幅 [sec]

sos = butter(N=order, Wn=[lowcut, highcut], btype='bandpass', fs=fs, output='sos')

def apply_filter_segmented(y, sos_mat):
    """NaN を含む 1‑D 配列にセグメントごとで sosfiltfilt を適用する"""
    good = np.isfinite(y)
    out  = np.full_like(y, np.nan)
    idx  = np.where(good)[0]
    segs = np.split(idx, np.where(np.diff(idx) != 1)[0] + 1)
    
    # パディング長はフィルタの次数に依存
    # scipyのドキュメントによると、sosfiltfiltのデフォルトpadlenは 3 * (sos.shape[1] // 2 - 1)
    # sosの形状は (n_sections, 6) なので、padlenは 3 * 2 = 6 となる
    padlen = 3 * (sos_mat.shape[1] - 1)
    
    for s in segs:
        if s.size > padlen:
            out[s] = sosfiltfilt(sos_mat, y[s])
    return out

B_64Hz_perp_bandpass = np.zeros(B_64Hz_perp.data.shape) * np.nan  # NaNで初期化
for i in range(3):
    B_64Hz_perp_bandpass[:, i] = apply_filter_segmented(B_64Hz_perp.data[:, i], sos)
da_B_64Hz_perp_bandpass = xr.DataArray(
    B_64Hz_perp_bandpass,
    dims=B_64Hz_perp.dims,
    coords=B_64Hz_perp.coords,
    name='B_64Hz_perp_bandpass'
)

da_B_64Hz_perp_bandpass_amp = np.sqrt((da_B_64Hz_perp_bandpass * da_B_64Hz_perp_bandpass).sum(dim='xyz'))
````

````python
def ensure_xyz_coord(da):
    if 'xyz' in da.dims and 'xyz' not in da.coords:
        da = da.assign_coords(xyz=['x','y','z'])
    return da

flux_pitch_zeta_data_list    = {}
for energy_i in range(energy_num):
    for channel_i in range(channel_num):
        v_unit  = v_unit_vector_dsi_energy_channel_list[energy_i, channel_i]
        B0 = B0_vector_dsi_energy_channel_list[energy_i, channel_i].interp(time=v_unit.time)
        Bperp = da_B_64Hz_perp_bandpass.interp(time=v_unit.time)
        Bperp = Bperp - (Bperp * B0).sum(dim='xyz') / (B0 * B0).sum(dim='xyz') * B0

        v_unit  = ensure_xyz_coord(v_unit)
        B0      = ensure_xyz_coord(B0)
        Bperp   = ensure_xyz_coord(Bperp)

        dot_vB0 = (v_unit * B0).sum(dim='xyz')
        B0_2    = (B0 * B0).sum(dim='xyz')
        alpha = np.arccos(dot_vB0 / np.sqrt(B0_2))

        v_perp = v_unit - dot_vB0 / B0_2 * B0
        cross = xr.apply_ufunc(np.cross, Bperp, v_perp,
                               input_core_dims=[['xyz'], ['xyz']],
                               output_core_dims=[['xyz']], vectorize=True)
        Bperp_2     = (Bperp * Bperp).sum(dim='xyz')
        sin_zeta    = (cross * B0).sum(dim='xyz') / np.sqrt(B0_2 * Bperp_2)
        cos_zeta    = (v_perp * Bperp).sum(dim='xyz') / np.sqrt(Bperp_2)
        zeta = np.atan2(sin_zeta, cos_zeta)
        
        # 時間を統一（zeta基準）
        t = zeta['time']
        
        # flux の time を zeta に合わせる（必要なら補間）
        flux_ch = xr.DataArray(
            flux_data_arrays[energy_i][:, channel_i],
            coords={'time': flux_data_arrays[energy_i].coords['time']},  # ここは実データのtimeに合わせる
            dims=('time',)
        ).interp(time=t)
        
        da = xr.concat(
            [
                flux_ch.rename('differential_number_flux_keV'),
                np.rad2deg(alpha).rename('pitch_angle_deg'),
                (np.rad2deg(zeta) % 360.0).rename('zeta_angle_deg')
            ],
            dim='variable'
        ).assign_coords(variable=['differential_number_flux_keV','pitch_angle_deg','zeta_angle_deg']) \
         .transpose('time','variable')
        
        flux_pitch_zeta_data_list[energy_i, channel_i] = da.assign_coords(
            channel=channel_ax[channel_i],
            energy_keV=energy_ax[energy_i],
        )
        print(flux_pitch_zeta_data_list[energy_i, channel_i])
````

````python
import numpy as np
from scipy.spatial import Delaunay

# --- 1) 時間窓の全点から分割を1回だけ作る ---
def build_triangulation_full(alpha_deg, zeta_deg, f):
    a = np.deg2rad(np.asarray(alpha_deg, float))
    z = np.deg2rad(np.asarray(zeta_deg,  float)) % (2*np.pi)
    f = np.asarray(f, float)
    m = np.isfinite(f)
    if m.sum() < 3:
        return None, None, None, None
    a, z, f = a[m], z[m], f[m]
    # ζの継ぎ目を滑らかにするため複製
    Z = np.concatenate([z-2*np.pi, z, z+2*np.pi])
    A = np.concatenate([a, a, a])
    F = np.concatenate([f, f, f])
    tri = Delaunay(np.column_stack([Z, A]))
    return tri, Z, A, F

# --- 2) 任意のζ-α矩形で重み付き積分（E/Bを切替） ---
def integrate_weighted_on_rect(tri, Z, A, F,
                               alpha_min_deg, alpha_max_deg,
                               zeta_min_deg,  zeta_max_deg,
                               mode="E",
                               n_samples=20000, rng=np.random):
    """
    mode='E': ∫∫ [-sinζ * sin^2α * f] dα dζ
    mode='B': ∫∫ [ cosζ * sin^2α * f] dα dζ
    Monte Carloで矩形Rに後から切り出し
    """
    if tri is None: return np.nan, 0.0
    if mode not in ("E","B"): raise ValueError("mode must be 'E' or 'B'.")

    amin = np.deg2rad(alpha_min_deg); amax = np.deg2rad(alpha_max_deg)
    z0 = np.deg2rad(zeta_min_deg) % (2*np.pi); z1 = np.deg2rad(zeta_max_deg) % (2*np.pi)
    if z1 <= z0: z1 += 2*np.pi

    # R内に一様サンプル
    zs  = z0  + (z1 - z0) * rng.random(n_samples)
    as_ = amin + (amax - amin) * rng.random(n_samples)

    pts = np.column_stack([zs, as_])
    simplex = tri.find_simplex(pts)
    valid = simplex >= 0
    if not np.any(valid): return np.nan, 0.0

    # バリセントリック補間
    X = pts[valid]
    T = tri.transform[simplex[valid]]                 # (M,3,2)
    delta = X - T[:,2]
    bary  = np.einsum('mij,mj->mi', T[:,:2], delta)   # (M,2)
    w0 = 1 - bary[:,0] - bary[:,1]; w1 = bary[:,0]; w2 = bary[:,1]
    verts = tri.simplices[simplex[valid]]             # (M,3)
    f_interp = w0*F[verts[:,0]] + w1*F[verts[:,1]] + w2*F[verts[:,2]]

    # 重み
    sin2a = np.sin(as_)**2
    w = (-np.sin(zs) if mode=="E" else np.cos(zs)) * sin2a

    g = w[valid] * f_interp
    area_R = (z1 - z0) * (amax - amin)
    I = area_R * np.nanmean(g)
    coverage = valid.mean()
    return I, coverage
````

````python
def _gather_samples_all(energy_i, tmin, tmax, variable):
    # αやζで切らない。±30 s内の全点を返す
    ts, al, ze, ff = [], [], [], []
    for ch in range(channel_num):
        d = modified_count_pitch_zeta_data_list[energy_i, ch].sel(time=slice(tmin, tmax))
        f = d.sel(variable=variable).values
        m = np.isfinite(f)
        if not m.any(): continue
        ts.append(d['time'].values[m])
        al.append(d.sel(variable='pitch_angle_deg').values[m])
        ze.append(d.sel(variable='zeta_angle_deg').values[m] % 360.0)
        ff.append(f[m])
    if not ts: return None, None, None
    return np.concatenate(al), np.concatenate(ze), np.concatenate(ff)

# --- 積分パラメータ ---
win_sec = 30.0
step_sec = 4.0
t0_list = np.arange(np.datetime64('2018-02-10T17:35:00'),
                    np.datetime64('2018-02-10T17:41:00') + np.timedelta64(1,'s'),
                    np.timedelta64(int(step_sec),'s'))
t0_list_ns = t0_list.astype('datetime64[ns]')

nT = len(t0_list)
alpha_min, alpha_max = 75., 105.

W_Eint_per_energy_list      = {}
W_Eint_per_energy_cov_list  = {}
W_Bint_per_energy_list      = {}
W_Bint_per_energy_cov_list  = {}

from joblib import Parallel, delayed
import numpy as np
import xarray as xr
import os

def compute_one_energy(energy_i):
    """
    1つの energy_i について、WEint_arr, covE_arr, WBint_arr, covB_arr を計算して返す。
    t0 方向はシーケンシャル（ネスト並列は避ける）。
    """
    WEint_arr = np.full(nT, np.nan, dtype=float)
    covE_arr  = np.zeros(nT, dtype=float)
    WBint_arr = np.full(nT, np.nan, dtype=float)
    covB_arr  = np.zeros(nT, dtype=float)

    for i, t0 in enumerate(t0_list):
        tmin = t0 - np.timedelta64(int(win_sec), 's')
        tmax = t0 + np.timedelta64(int(win_sec), 's')

        # --- E ---
        a_E, z_E, f_E = _gather_samples_all(energy_i, tmin, tmax, 'modified_count_E')
        if a_E is not None:
            triE, ZE, AE, FE = build_triangulation_full(a_E, z_E, f_E)
            if triE is not None:
                IE, covE = integrate_weighted_on_rect(
                    triE, ZE, AE, FE,
                    alpha_min, alpha_max,
                    0.0, 360.0,
                    mode="E", n_samples=20000
                )
                WEint_arr[i] = IE * energy_width_center[energy_i - 1]
                covE_arr[i]  = covE

        # --- B ---
        a_B, z_B, f_B = _gather_samples_all(energy_i, tmin, tmax, 'modified_count_B')
        if a_B is not None:
            triB, ZB, AB, FB = build_triangulation_full(a_B, z_B, f_B)
            if triB is not None:
                IB, covB = integrate_weighted_on_rect(
                    triB, ZB, AB, FB,
                    alpha_min, alpha_max,
                    0.0, 360.0,
                    mode="B", n_samples=20000
                )
                WBint_arr[i] = IB * energy_width_center[energy_i - 1]
                covB_arr[i]  = covB

    # xarray化（この関数の戻り値でまとめて受け取る）
    da_WE = xr.DataArray(
        WEint_arr, dims=['time'],
        coords={'time': t0_list_ns,
                'energy_center_keV': energy_center[energy_i - 1],
                'energy_width':      energy_width[energy_i - 1]},
        name=f'W_Eint_per_energy_{energy_i}'
    )
    da_WE_cov = xr.DataArray(
        covE_arr, dims=['time'],
        coords={'time': t0_list_ns,
                'energy_center_keV': energy_center[energy_i - 1],
                'energy_width':      energy_width[energy_i - 1]},
        name=f'W_Eint_per_energy_cov_{energy_i}'
    )
    da_WB = xr.DataArray(
        WBint_arr, dims=['time'],
        coords={'time': t0_list_ns,
                'energy_center_keV': energy_center[energy_i - 1],
                'energy_width':      energy_width[energy_i - 1]},
        name=f'W_Bint_per_energy_{energy_i}'
    )
    da_WB_cov = xr.DataArray(
        covB_arr, dims=['time'],
        coords={'time': t0_list_ns,
                'energy_center_keV': energy_center[energy_i - 1],
                'energy_width':      energy_width[energy_i - 1]},
        name=f'W_Bint_per_energy_cov_{energy_i}'
    )
    return energy_i, da_WE, da_WE_cov, da_WB, da_WB_cov

if __name__ == "__main__":
    # 並列実行（energy_i=1..29）
    n_jobs = os.cpu_count()
    results = Parallel(n_jobs=n_jobs, backend="loky", verbose=5)(
        delayed(compute_one_energy)(energy_i)
        for energy_i in range(1, energy_num)
    )

    # 受け取り & 連結
    W_Eint_per_energy_list      = {}
    W_Eint_per_energy_cov_list  = {}
    W_Bint_per_energy_list      = {}
    W_Bint_per_energy_cov_list  = {}

    for energy_i, da_WE, da_WE_cov, da_WB, da_WB_cov in results:
        W_Eint_per_energy_list[energy_i]      = da_WE
        W_Eint_per_energy_cov_list[energy_i]  = da_WE_cov
        W_Bint_per_energy_list[energy_i]      = da_WB
        W_Bint_per_energy_cov_list[energy_i]  = da_WB_cov

    W_Eint_all = xr.concat(list(W_Eint_per_energy_list.values()), dim='energy_center_keV')
    W_Eint_cov_all = xr.concat(list(W_Eint_per_energy_cov_list.values()), dim='energy_center_keV')
    W_Bint_all = xr.concat(list(W_Bint_per_energy_list.values()), dim='energy_center_keV')
    W_Bint_cov_all = xr.concat(list(W_Bint_per_energy_cov_list.values()), dim='energy_center_keV')

    W_Eint_all.name = 'W_Eint'; W_Eint_all.attrs['units'] = '[eV m^-3]'
    W_Eint_cov_all.name = 'W_Eint_cov'
    W_Bint_all.name = 'W_Bint'; W_Bint_all.attrs['units'] = '[eV s m^-4]'
    W_Bint_cov_all.name = 'W_Bint_cov'

    print(W_Eint_all); print(W_Eint_cov_all)
    print(W_Bint_all); print(W_Bint_cov_all)
````

````python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

def plot_W_on_ax(ax, W_all, energy_range=None, time_range=None,
                 cmap="turbo", normalize=True, vlim=None,
                 title=None, ylabel="Energy [keV]", cbar_title=None, cax=None):
    Wa = W_all

    # --- energy 範囲: 降順でも動くようにブール抽出 ---
    if energy_range is not None:
        el, eh = energy_range
        ec = Wa.coords["energy_center_keV"].values
        mask = (ec >= min(el, eh)) & (ec <= max(el, eh))
        Wa = Wa.isel(energy_center_keV=mask)

    # --- time 範囲 ---
    if time_range is not None:
        t0, t1 = np.datetime64(time_range[0]), np.datetime64(time_range[1])
        Wa = Wa.sel(time=slice(t0, t1))

    if Wa.size == 0:
        raise ValueError("指定範囲内にデータが存在しない")

    # --- energy 昇順に並べ替え ---
    e = Wa.energy_center_keV.values
    order = np.argsort(e)
    e = e[order]
    Wa = Wa.isel(energy_center_keV=order)

    # --- energy エッジ（対数） ---
    loge = np.log10(e)
    loge_edge = np.empty(e.size + 1)
    loge_edge[1:-1] = 0.5 * (loge[:-1] + loge[1:])
    loge_edge[0]    = loge[0]  + (loge[0]  - loge_edge[1])
    loge_edge[-1]   = loge[-1] - (loge_edge[-2] - loge[-1])
    e_edge = 10**loge_edge

    # --- time エッジ ---
    t = Wa.time.values
    t_num = mdates.date2num(t.astype("datetime64[ms]").astype(object))
    if t_num.size == 1:
        dt = 1/24/60
        t_edge = np.array([t_num[0]-dt/2, t_num[0]+dt/2])
    else:
        t_edge = np.empty(t_num.size + 1)
        t_edge[1:-1] = 0.5*(t_num[:-1] + t_num[1:])
        t_edge[0]    = t_num[0]  - (t_edge[1]  - t_num[0])
        t_edge[-1]   = t_num[-1] + (t_num[-1] - t_edge[-2])

    Z = Wa.values

    # --- 正規化 or vlim ---
    if normalize:
        if vlim is None:
            vmax = np.nanmax(np.abs(Z)) or 1.0
            Z = Z / vmax
            vmin, vmax_plot = -1, 1
        else:
            vmax = np.nanmax(np.abs(Z)) or 1.0
            Z = Z / vmax
            vmin, vmax_plot = vlim
    else:
        if vlim is None:
            vmax_plot = np.nanmax(np.abs(Z)) or 1.0
            vmin = -vmax_plot
        else:
            vmin, vmax_plot = vlim

    pcm = ax.pcolormesh(t_edge, e_edge, Z, shading="auto",
                        cmap=cmap, vmin=vmin, vmax=vmax_plot)
    ax.set_yscale("log")
    ax.set_ylabel(ylabel)
    ax.minorticks_on()
    ax.grid(True, which='both', linestyle='--', alpha=0.5)
    ax.xaxis.set_major_locator(mdates.MinuteLocator(interval=1))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
    if title: ax.set_title(title)
    if cax is not None:
        cb = plt.colorbar(pcm, cax=cax)
        if cbar_title: cb.set_label(cbar_title)
    return pcm
````

````python
import numpy as np
import xarray as xr

def integrate_weighted_energy_sum(alpha_min, alpha_max,
                                  zeta_min, zeta_max,
                                  E_min, E_max,
                                  mode, tmin, tmax,
                                  n_samples=20000):
    """
    各 energy_i ごとに ζ–α 積分を行い ΔE/E で加重和。
    """
    total_I = 0.0
    cov_list = []

    for energy_i in range(energy_num):
        if energy_i == 0:
            continue
        Ecen = energy_center[energy_i-1]
        if not (E_min <= Ecen <= E_max):
            continue

        # modified_count_E/B から取得
        var = 'modified_count_E' if mode == "E" else 'modified_count_B'
        a, z, f = _gather_samples_all(energy_i, tmin, tmax, var)
        if a is None:
            continue

        tri, Z, A, F = build_triangulation_full(a, z, f)
        if tri is None:
            continue

        I2d, cov = integrate_weighted_on_rect(
            tri, Z, A, F,
            alpha_min_deg=alpha_min,
            alpha_max_deg=alpha_max,
            zeta_min_deg=zeta_min,
            zeta_max_deg=zeta_max,
            mode=mode,
            n_samples=n_samples
        )

        total_I += (energy_width[energy_i] / Ecen) * I2d
        cov_list.append(cov)

    cov_mean = np.nanmean(cov_list) if cov_list else 0.0
    return total_I, cov_mean


def compute_time_series_EB_energy_sum(alpha_min=75., alpha_max=105.,
                                      zeta_min=0., zeta_max=360.,
                                      E_min=0.026, E_max=0.725,
                                      win_sec=30.0, step_sec=4.0,
                                      n_samples=20000,
                                      n_jobs=None):
    """
    各時刻ごとに ζ–α–E 積分を行い、結果を xarray.DataArray(time) に格納。
    """
    t0_list = np.arange(np.datetime64('2018-02-10T17:35:00'),
                        np.datetime64('2018-02-10T17:41:00') + np.timedelta64(1, 's'),
                        np.timedelta64(int(step_sec), 's'))
    t0_list_ns = t0_list.astype('datetime64[ns]')
    nT = len(t0_list)
    n_jobs = n_jobs or os.cpu_count()

    def process_one_time(t0):
        tmin = t0 - np.timedelta64(int(win_sec), 's')
        tmax = t0 + np.timedelta64(int(win_sec), 's')

        IE, covE = integrate_weighted_energy_sum(
            alpha_min, alpha_max, zeta_min, zeta_max,
            E_min, E_max, mode="E",
            tmin=tmin, tmax=tmax,
            n_samples=n_samples
        )
        IB, covB = integrate_weighted_energy_sum(
            alpha_min, alpha_max, zeta_min, zeta_max,
            E_min, E_max, mode="B",
            tmin=tmin, tmax=tmax,
            n_samples=n_samples
        )
        return IE, covE, IB, covB

    results = Parallel(n_jobs=n_jobs, backend="loky", verbose=5)(
        delayed(process_one_time)(t0) for t0 in t0_list
    )

    WE_arr, covE_arr, WB_arr, covB_arr = map(np.array, zip(*results))

    da_WE = xr.DataArray(
        WE_arr * (win_sec * 2.), dims=['time'], coords={'time': t0_list_ns},    # 2*win_secで時間積分
        name='W_Eint_energy_sum', attrs={'units': '[eV m^-3]'}
    )
    da_WB = xr.DataArray(
        WB_arr * (win_sec * 2.), dims=['time'], coords={'time': t0_list_ns},
        name='W_Bint_energy_sum', attrs={'units': '[eV s m^-4]'}
    )
    da_covE = xr.DataArray(
        covE_arr, dims=['time'], coords={'time': t0_list_ns},
        name='W_Eint_cov'
    )
    da_covB = xr.DataArray(
        covB_arr, dims=['time'], coords={'time': t0_list_ns},
        name='W_Bint_cov'
    )

    return da_WE, da_covE, da_WB, da_covB
````
